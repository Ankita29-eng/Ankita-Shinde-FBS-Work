### Q8.Plot a heatmap to represent entire dataset's correlation

```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
df=pd.read_csv("C:/Users/Ankita shinde/OneDrive/Desktop/FBS/Data Analysis 1/DataSets/Datasets/tips.csv")

correlation = df.corr(numeric_only=True)

plt.figure(figsize=(8,6))

sns.heatmap(correlation,annot=True,cmap="coolwarm")

plt.title("Correlation Heatmap of Tips Dataset")
plt.show()
```
