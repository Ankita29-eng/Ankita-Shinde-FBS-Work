### 5.  Line Chart:


####  Create a line chart showing the trend of horsepower for different types of cars.
####  Include proper legends, labels, and a title

```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("C:/Users/Ankita shinde/OneDrive/Desktop/FBS/Data Analysis 1/DataSets/Datasets/cars93.csv")

avg_hp = df.groupby("Type")["Horsepower"].mean()

plt.figure(figsize=(10,6))

plt.plot(avg_hp.index,avg_hp.values,marker="o",label="Average Horsepower")

plt.xlabel("Car Type")
plt.ylabel("Average Horsepower")
plt.title("Trend of Horsepower for Different Types of Cars")

plt.legend()

plt.show()
```
