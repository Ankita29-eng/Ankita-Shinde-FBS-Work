### Q6. Draw a scatter plot of tip and totalbill and also allot the points according to the sex

```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("C:/Users/Ankita shinde/OneDrive/Desktop/FBS/Data Analysis 1/DataSets/Datasets/tips.csv")

plt.figure(figsize=(8,6))

for sex in df["sex"].unique():
    data = df[df["sex"]==sex]
    plt.scatter(data["total_bill"],data["tip"],label=sex)

plt.xlabel("Total bill")
plt.ylabel("Tip")
plt.title("Relationship between Total Bill and Tip by Sex")

plt.legend()

plt.show()
```
