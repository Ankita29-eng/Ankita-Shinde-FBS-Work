### 4. Scatter Plot:


#### a. Create a scatter plot to show the relationship between horsepower and MPG (Miles Per Gallon).
#### b. Label the axes and add a title.

```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("C:/Users/Ankita shinde/OneDrive/Desktop/FBS/Data Analysis/DataSets/Datasets/cars93.csv")

plt.scatter(df["Horsepower"],df["MPG.city"])

plt.xlabel("Horsepower")
plt.ylabel("MPG (Miles Per Gallon)")
plt.title("Relationship between Horsepower and MPG")

plt.show()
```
