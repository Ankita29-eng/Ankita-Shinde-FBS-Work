### 3. Pie Chart:


#### a. Create a pie chart showing the proportion of different car types in the dataset.
#### b. Include percentages in the chart.

```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("C:/Users/Ankita shinde/OneDrive/Desktop/FBS/Data Analysis/DataSets/Datasets/cars93.csv")

type_count=df["Type"].value_counts()

plt.figure(figsize=(8,8))
plt.pie(type_count.values,
labels=type_count.index,
autopct="%1.1f%%")

plt.title("Proportion of Different Car Types")
plt.show()
```
