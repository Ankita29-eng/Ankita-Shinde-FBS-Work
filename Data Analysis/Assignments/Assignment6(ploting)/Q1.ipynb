### 1. Bar Chart:


#### a. Create a bar chart to compare the average price of cars by manufacturer.
#### b. Label the axes and add a suitable title.
#### c. Use different colors for different manufacturers.

```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("C:/Users/Ankita shinde/OneDrive/Desktop/FBS/Data Analysis/DataSets/Datasets/cars93.csv")

avg_price = df.groupby("Manufacturer")["Price"].mean()

plt.figure(figsize=(12,6))
plt.bar(avg_price.index,avg_price.values,color=plt.cm.tab20.colors[:len(avg_price)])

plt.xlabel("Manufacturer")
plt.ylabel("Average Price")
plt.title("Average Price of Cars by Manufacturer")
plt.xticks(rotation=90)

plt.show()
```
