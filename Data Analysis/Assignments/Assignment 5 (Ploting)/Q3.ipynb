#### Q3.Draw a scatter plot of tip and totalbill and also allot the points according to the sex

```python
import pandas as pd
```

```python
import seaborn as sns
```

```python
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("C:/Users/Ankita shinde/OneDrive/Desktop/FBS/Data Analysis/DataSets/Datasets/tips.csv")
sns.scatterplot(x="total_bill",y="tip",hue="sex",data=df)
plt.title("Scatter Plot of Tip and Total Bill")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.show()
```
