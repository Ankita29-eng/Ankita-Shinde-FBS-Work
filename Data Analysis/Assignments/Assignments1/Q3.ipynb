#### Q3.Perform the following operations on an array of mobile phones prices 6999,7500,11999,27899,14999,9999

##### a.Create a 1d-array of mobile phones prices

```python
import numpy as np
```

```python
arr = np.array([6999,7500,11999,27899,14999,9999])
print("Original Array:")
print(arr)
```

##### b.Convert this array to float type

```python
arr=arr.astype(float)
print("\nFloat Array:")
print(arr)
```

##### c.Append a new mobile having price of 13999 Rs. to this array.

```python
arr = np.append(arr,13999)
print("\n After Appending 13999:")
print(arr)
```

##### d.Reverse this array of mobile phones prices

```python
arr = arr[::-1]
print("\nReversed Array:")
print(arr)
```

##### e.Apply GST of 18% on mobile phones prices and update this array

```python
arr = arr*1.18
print("\nAfter applying 18% GST:")
print(arr)
```

##### f.Sort the array in descending order of price

```python
arr = np.sort(arr)[::-1]
print("\nSorted Array in Descending Order")
print(arr)
```

##### g.What is the average mobile phone price

```python
average = np.mean(arr)
print("\nAverage Mobile Phone Price:",average)
```

##### h.What is the difference b/w maximum and minimum price

```python
difference=np.max(arr)-np.min(arr)
print("Difference between Maximum and Minimum Price:",difference)
```
