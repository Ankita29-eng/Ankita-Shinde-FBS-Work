##### Q2.Write code to replace the odd numbers by -1 in the array created by using random.randint(1,50,20)

```python
import numpy as np
```

```python
arr = np.random.randint(1,50,20)
print("Original Array:")
print(arr)
arr[arr % 2!=0]=-1
print("\nArray after replacing odd numbers with -1:")
print(arr)
```
