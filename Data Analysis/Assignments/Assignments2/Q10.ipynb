#### Q10.Create a 2x2 matrix A and a 2x2 matrix B.Perform matrix multiplication using np.dot() or the @ operator.

```python
import numpy as np
```

```python
A = np.array([[1,2],
              [3,4]])
```

```python
B= np.array([[5,6],
             [7,8]])
```

```python
print("Matrix A:")
print(A)
```

```python
print("Matrix B:")
print(B)
```

```python
print("Matrix Multiplication:")
print(np.dot(A,B))
```
