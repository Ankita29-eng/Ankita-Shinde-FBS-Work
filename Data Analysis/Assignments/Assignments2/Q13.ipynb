#### Q13.Create a 4x4 array of random integers.Find the maximum,minimum,and mean of the entire array.

```python
import numpy as np
```

```python
arr = np.random.randint(1,101,size = (4,4))
```

```python
print("Array:")
print(arr)
```

```python
print("Maximum:",np.max(arr))
```

```python
print("Minimum:",np.min(arr))
```

```python
print("Mean:",np.mean(arr))
```
