#### Q8.Create two arrays of shape (2,3): arr1 and arr2. Perform element-wise addition,subtraction,multiplication,and division on them.

```python
import numpy as np
```

```python
arr1 =np.array([[10,20,30],
                [40,50,60]])
```

```python
arr2 = np.array([[2,4,5],
                 [5,10,20]])
```

```python
print("Addition:")
print(arr1+arr2)
```

```python
print("Subtraction:")
print (arr1 - arr2)
```

```python
print("Multiplication:")
print(arr1 * arr2)
```

```python
print("Division:")
print(arr1/arr2)
```
