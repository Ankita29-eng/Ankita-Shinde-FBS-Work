#### Q11.Find the transpose of matrix A and matrix B.

```python
import numpy as np
```

```python
A = np.array([[1,2],
              [3,4]])
```

```python
B = np.array([[5,6],
             [7,8]])
```

```python
print("Transpose of Matrix A:")
print(A.T)
```

```python
print("Transpose of Matrix B:")
print(B.T)
```
