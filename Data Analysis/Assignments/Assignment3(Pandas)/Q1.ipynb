##### d.Display total number of apps which even supports Andriod version 2.3 and up

```python
df['Android Ver'] = df['Android Ver'].str.extract(r'(\d+(?:\.\d+)?)')[0]

df['Android Ver'] = pd.to_numeric(df['Android Ver'], errors='coerce')

count = (df['Android Ver'] >= 2.3).sum()

print(count)
```
