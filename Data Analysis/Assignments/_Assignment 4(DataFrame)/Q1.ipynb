### 1. Create a DataFrame:
#### o Load the above dataset into a Pandas DataFrame.

```python
import pandas as pd
```

```python
data = {
    "Order ID": [1001, 1002, 1003, 1004, 1005, 1006, 1007, 1008],
    "Customer": ["Alice", "Bob", "Charlie", "Alice", "David", "Charlie", "Bob", "Alice"],
    "Product": ["Laptop", "Phone", "Chair", "Table", "Laptop", "Phone", "Chair", "Phone"],
    "Category": ["Electronics", "Electronics", "Furniture", "Furniture",
                 "Electronics", "Electronics", "Furniture", "Electronics"],
    "Quantity": [2, 1, 4, 1, 1, 2, 2, 1],
    "Price": [800, 500, 100, 250, 800, 500, 100, 500],
    "Total Sales": [1600, 500, 400, 250, 800, 1000, 200, 500]
}

df = pd.DataFrame(data)
print(df)
```

### 2. Add a New Column:
#### o Add a column Discount that gives a 10% discount on total sales.

```python
df["Discount"] = df["Total Sales"] * 10 / 100

print(df)
```

### 3. Add a New Row:
#### o Add a new sales record: Order ID = 1009, Customer ="Eve", Product = "Tablet", Category = "Electronics", Quantity = 1, Price = 300, Total Sales = 300.

```python
df.loc[len(df)]=[1009,"Eve","Tablet","Electronics",1,300,300,30]
print(df)
```

### 4. Filter Data:
#### o Retrieve only the sales records where the Category is "Electronics".

```python
electronics = df[df["Category"]=="Electronics"]
print(electronics)
```

### 5. Selection:
#### o Select the Customer and Total Sales columns.

```python
selected = df[["Customer","Total Sales"]]
print(selected)
```

### 6. Grouping:
#### o Group the data by Customer and calculate the total sales per customer.

```python
grouped = df.groupby("Customer")["Total Sales"].sum()
print(grouped)
```

### 7. Sorting:
#### o Sort the DataFrame based on Total Sales in descending order.

```python
sorted_df = df.sort_values(by="Total Sales",ascending=False)
print(sorted_df)
```

### 8. Apply Function:
#### o Apply a lambda function to create a new column Sales Category, where values are labeled as "High" if Total Sales > 1000, otherwise "Low".

```python
df["Sales Category"]=df["Total Sales"].apply(lambda x:"High" if x>1000 else "Low")
print(df)
```

### 9. Save DataFrame:
#### o Save the final DataFrame as a CSV file named sales_data.csv.

```python
df.to_csv("sales_data.csv",index=False)
```
